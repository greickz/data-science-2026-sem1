```python
import pandas as pd
df = pd.read_excel('faltas_treinamento_sujo.xlsx')
print(df)
```

Exercicio 1

A)

```python
print(df.isnull().sum())
print(df.dtypes)
```

B)

```python
df['Data_Ultima_Falta'] = pd.to_datetime(df['Data_Ultima_Falta'], errors='coerce')
data_substituta = pd.Timestamp("2024-02-29")
df['Data_Ultima_Falta'] = df['Data_Ultima_Falta'].fillna(data_substituta)
print(df.isnull().sum())
print(df)
```

C)

```python
df['Turno'] = df['Turno'].str.title().replace({'manha':'Manhã','Manha': 'Manhã'})
df['Dia_da_Semana_Ultima_Falta'] = df['Dia_da_Semana_Ultima_Falta'].str.title()

print(df)
```

Exercicio 2

A)

```python
df = df.drop_duplicates(subset=['Nome'])
print(df)
```

B)

```python
df['Evadiu'].fillna('Desconhecida', inplace=True)
df.head(50)
```

C)

```python
data_mais_comum = df['Data_Ultima_Falta'].mode()[0]
df['Data_Ultima_Falta'] = df['Data_Ultima_Falta'].fillna(data_mais_comum)
print(df)
```

Exercicio 3

A)

```python
df['Faltas'] = pd.to_numeric(df['Faltas'], errors='coerce').astype('Int64')
print(df.dtypes)
```

B)

```python
import numpy as np
df['Faltas'] = pd.to_numeric(df['Faltas'], errors='coerce')
media = df['Faltas'].mean()
desvio = df['Faltas'].std()
outlier_tratamento = (df['Faltas'] < media - 2 * desvio) | (df['Faltas'] > media + 2 * desvio)
df.loc[outlier_tratamento, 'Faltas'] = np.nan
df['Faltas'] = df['Faltas'].fillna(df['Faltas'].median()).astype(int)
print(df)
```

C)

```python
df['Faltas_Suspeitas'] = df['Faltas'] > 10
print(df) 
```

```python
df.to_excel('planilha_atualizada.xlsx', index=False)
```
