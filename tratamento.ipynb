```python
import pandas as pd
df = pd.read_excel(" .xlsx")
```

```python
# DIVIDIR EM TEXTOS E NÚMEROS

texto_separado = df.select_dtypes(include="object")
numero_separado = df.select_dtypes(include="number")
```

```python
# ORGANIZAR OS VALORES EM FLOAT

```

```python
# Captalize() e Strip()

df[""]=df[""].str.capitalize()
df[""]=df[""].str.strip()
```

```python
# Remove o "R$"
df['valor'] = df['valor'].str.replace('R$', '', regex=False)
print(df)
```

```python
# Remove o "."
df['valor'] = df['valor'].str.replace('.', '', regex=False)
print(df)
```

```python
# Substitui a vírgula (',') pelo ponto ('.') nas casas decimais 
df['valor'] = df['valor'].str.replace(',', '.', regex=False)
print(df)
```

```python
# Verifica se há espaços entre os caracteres e transforma o tipo em float
df['valor'] = df['valor'].str.strip().astype(float)
print(df)
```

```python
# Gerar o excell com os dados tratados
df.to_excel('dados_tratados_vivareal.xlsx', index=False)
```
